```python
from pathlib import Path
import pandas as pd

base = Path("dados_resumidos")

for year in [2022, 2023, 2024]:
    path = base / f"tmi_municipios_{year}.parquet"
    df = pd.read_parquet(path)

    print("\n" + "="*80)
    print(f"{year} | shape = {df.shape}")
    print("COLUNAS:")
    print(list(df.columns))

    # opcional: ver 3 linhas pra entender conteúdo
    print("\nHEAD (3 linhas):")
    print(df.head(3))
```
